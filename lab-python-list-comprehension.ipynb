````python
products = ["t-shirt", "mug", "hat", "book", "keychain"]
````

````python
def initialize_inventory(products):
    inventory = {product: int(input(f"Enter the quantity of {product}s available: ")) for product in products}
    return inventory   
````

````python
def get_customer_orders():
    number_orders = int(input("Enter the number of customer orders: "))
    
    customer_orders = {
        input("Enter a product you want to order: ")
        for i in range(number_orders)
    } 
    return customer_orders
````

````python
def calculate_total_price(customer_orders):
    total_price = sum( 
        float(input(f"Enter the price of {product}: "))
        for product in customer_orders
    )

    return total_price 
````

````python
def update_inventory(customer_orders, inventory):
    for product in customer_orders:
        inventory[product] = inventory[product] - 1

    filtered_inventory = {
        product: quantity
        for product, quantity in inventory.items()
        if quantity > 0 
    }     

    return filtered_inventory
````

````python
def calculate_order_statistics(customer_orders, products):
    total_products_ordered = len(customer_orders)
    percentage = (total_products_ordered / len(products)) * 100

    return total_products_ordered, percentage
````

````python
def print_order_statistics(total_products_ordered, percentage):
    print()
    print("Order Statistics:")
    print(f"Total Products Ordered: {total_products_ordered}")
    print(f"Percentage of Unique Products Ordered: {percentage}")
````

````python
inventory = initialize_inventory(products)
customer_orders = get_customer_orders()

total_products_ordered, percentage = calculate_order_statistics(
    customer_orders,
    products
)

print_order_statistics(total_products_ordered, percentage)

inventory = update_inventory(customer_orders, inventory)

print()
print("Updated Inventory:")

[print(f"{product}: {quantity}") for product, quantity in inventory.items()]

total_price = calculate_total_price(customer_orders)

print(f"Total Price: {total_price}")
````
